```python
import pandas as pd

ipl = pd.read_csv(
    r"C:\Users\hp\Downloads\IPL.csv",
    low_memory=False
)
```

```python
print(ipl.columns.tolist())
```

```python
toss_wins = ipl['toss_winner'].value_counts()

print(toss_wins.head())
```

```python
toss_wins.head(5).plot(kind='pie', autopct='%1.1f%%')

plt.title('Top Toss Winning Teams')

plt.ylabel('')

plt.show()
```

```python
season_matches = ipl['season'].value_counts().sort_index()

print(season_matches)
```

```python
season_matches.plot(kind='line', marker='o')

plt.title('IPL Matches Per Season')
plt.xlabel('Season')
plt.ylabel('Number of Matches')

plt.show()
```

```python
print(ipl.columns.tolist())
```

```python
venue_matches = ipl['venue'].value_counts()

print(venue_matches.head())
```

```python
import matplotlib.pyplot as plt

venue_matches.head(10).plot(
    kind='barh',
    figsize=(10,6)
)

plt.title('Top IPL Venues')
plt.xlabel('Number of Matches')
plt.ylabel('Venue')

plt.show()
```

```python
best_players = ipl['player_of_match'].value_counts()

print(best_players.head())
```

```python
import matplotlib.pyplot as plt

best_players.head(10).plot(
    kind='bar',
    figsize=(10,5)
)

plt.title('Top IPL Players')
plt.xlabel('Players')
plt.ylabel('Player of Match Awards')

plt.xticks(rotation=45)

plt.show()
```
